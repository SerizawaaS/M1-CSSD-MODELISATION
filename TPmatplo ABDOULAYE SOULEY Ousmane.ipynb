## Exercise 1: Simple Polynomial Plot

1. Plot a Polynomial Function:
   - Create an array `x` with 100 linearly spaced points between -10 and 10.

### Exercise 6: Contour Plot

1. Plot a Contour Plot of a Mathematical Function:
   - Create an array `x` of 200 points linearly spaced between -5 and 5, and a similar array `y`.
   - Create a 2D grid with `numpy.meshgrid` and define the function `f(x, y) = sin(sqrt(x^2 + y^2))`.
   - Plot a contour plot of this function.
2. Customization:
   - Add labels to the contours and use a different colormap.

```python
import numpy as np
import matplotlib.pyplot as plt

# Definir X Y
x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)

# TRANSFORMATION de x y en matrices 2D
X, Y = np.meshgrid(x, y)

# DEFINIR LA FONCTION
def f(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

Z = f(X, Y)

# definition de l'écran d'affichage
plt.figure(figsize=(7, 6))  # Adjust figure size

# remplir le contour
contour_filled = plt.contourf(X, Y, Z, levels=20, cmap='plasma', alpha=0.75)

# lignes
contour_lines = plt.contour(X, Y, Z, levels=8, colors='black')

# ajouter les labels aux contours
plt.clabel(contour_lines, inline=True, fontsize=10)

# rajouter la barre de couleurs 
plt.colorbar(contour_filled)

# Rajouter titres et nom d'axes
plt.title("Contour de f(x, y) = sin(sqrt(x² + y²))")
plt.xlabel("X")
plt.ylabel("Y")

# Afficher
plt.show()
```

## Exercise 7: 3D Surface Plot

1. Plot a 3D Surface Plot of a Mathematical Function:
   - Create arrays `X` and `Y` with values ranging from -5 to 5 with a step of 0.25.
   - Compute `Z = cos(sqrt(X^2 + Y^2))`.
   - Plot a 3D surface plot of `Z`.
2. Customization:
   - Change the colormap and the viewing angle of the plot. Add a color bar.

```python
#Création de la figure
fig=plt.figure(figsize=(30,12),dpi=80)
ax=fig.add_subplot(111,projection='3d')
#création de X et Y
X = np.arange(-5, 5.25, 0.25)  # 5.25 pour include 5 dans le range
Y = np.arange(-5, 5.25, 0.25)
X,Y=np.meshgrid(X,Y)
#création de la surface 3d
Z=np.cos(np.sqrt(X**2+Y**2))
ax.plot_surface(X, Y, Z, rstride=1, cstride=1, cmap=plt.cm.hot)
ax.contourf(X, Y, Z, zdir='z', offset=-2, cmap=plt.cm.hot)

# LIMITES POUR Z
ax.set_zlim(-3, 3)

plt.show()
```

## Exercise 10: Changing Viewing Angle

1. Change the Viewing Angle of a 3D Plot:
   - Reuse the 3D surface plot from Exercise 7.
   - Create two subplots with different viewing angles using `view_init`.
2. Customization:
   - Add titles to each viewing angle.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Création de X et Y
X = np.arange(-5, 5.25, 0.25)  # 5.25 pour inclure 5 dans le range
Y = np.arange(-5, 5.25, 0.25)
X, Y = np.meshgrid(X, Y)

# Création de la surface 3D
Z = np.cos(np.sqrt(X**2 + Y**2))

# Création de la figure avec deux sous-graphiques
fig, axes = plt.subplots(1, 2, figsize=(12, 5), subplot_kw={"projection": "3d"})

# Premier angle de vue
axes[0].plot_surface(X, Y, Z, cmap="viridis")
axes[0].set_title("Vue de face")
axes[0].view_init(elev=30, azim=45)  # Réglage de l'angle

# Deuxième angle de vue
axes[1].plot_surface(X, Y, Z, cmap="plasma")
axes[1].set_title("Vue latérale")
axes[1].view_init(elev=20, azim=180)  # Un autre angle de vue

# Affichage de la figure
plt.show()

```

### Exercise 15: 3D Vector Field

1. Plot a 3D Vector Field:
   - Create arrays `X`, `Y`, and `Z` with values ranging from -5 to 5.
   - Compute `U = -Y`, `V = X`, and `W = Z`.
   - Plot a 3D vector field using `quiver3D`.
2. Customization:
   - Add a title and labels to the axes. Change the color and length of the vectors.

```python
from mpl_toolkits.mplot3d import Axes3D
# Création des grilles de points X, Y, Z (de -5 à 5)
x = np.linspace(-5, 5, 10)  # 10 points entre -5 et 5
y = np.linspace(-5, 5, 10)
z = np.linspace(-5, 5, 5)   # Moins de points en Z pour éviter un excès de vecteurs
# Création de la grille 3D
X, Y, Z = np.meshgrid(x, y, z)
# Définition des composantes du champ de vecteurs
U = -Y   # Composante en X
V = X    # Composante en Y
W = Z    # Composante en Z (ici, identique à Z pour illustrer une montée)
# Création de la figure et de l'axe 3D
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')
# Tracé du champ de vecteurs en 3D avec `quiver`
ax.quiver(X, Y, Z, U, V, W, length=0.8, color='blue', linewidth=1)
# Personnalisation des axes
ax.set_xlabel("Axe X")
ax.set_ylabel("Axe Y")
ax.set_zlabel("Axe Z")
ax.set_title("Champ de vecteurs 3D")
# Affichage du graphique
plt.show()
```

### Exercise 20: 3D Density Plot

1. Plot a 3D Density Plot:
   - Generate 1000 random points in 3D space.
   - Create a density plot to show where points are concentrated.
2. Customization:
   - Add labels to the axes. Adjust the transparency of the plot.

```python
from scipy.stats import gaussian_kde
#Génération de 1000 points aléatoires en 3D suivant une distribution normale
x = np.random.randn(1000)
y = np.random.randn(1000)
z = np.random.randn(1000)
#Calcul de la densité des points avec une estimation par noyau gaussien
xyz = np.vstack([x, y, z])
density = gaussian_kde(xyz)(xyz)
#Création de la figure et de l'axe 3D
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
#Tracé des points avec couleur en fonction de la densité et transparence ajustée
sc = ax.scatter(x, y, z, c=density, cmap='viridis', alpha=0.6)
#Ajout d'une barre de couleur pour la densité
colorbar = fig.colorbar(sc, ax=ax, shrink=0.5, aspect=10)
colorbar.set_label("Densité des points")
#Personnalisation des axes
ax.set_xlabel("Axe X")
ax.set_ylabel("Axe Y")
ax.set_zlabel("Axe Z")
ax.set_title("Nuage de points 3D avec densité")
#Affichage du graphique
plt.show()
```
