## Exercice 7 : Grouping and Aggregation

Group the following DataFrame by column Category and calculate the mean of column Value:

```python
   Category  Value
0         A      1
1         B      2
2         A      3
3         B      4
4         A      5
5         B      6
```

```python
df=pd.DataFrame(data={'Catégorie':['A','B','A','B','A','B'],'Value':[1,2,3,4,5,6]})
df
```

```python
#Calcul de l'effectif de chaque catégorie
df.groupby('Catégorie')['Catégorie'].value_counts()
```

```python
#calcul de la moyenne des valeurs pour chaque catégorie
df.groupby('Catégorie')['Value'].mean()
```

## Exercice 8 : Pivot Tables

Create a pivot table from the following DataFrame, showing the mean Value for each Category and Type:

```python
   Category  Type  Value
0         A     X      1
1         A     Y      2
2         A     X      3
3         B     Y      4
4         B     X      5
5         B     Y      6
```

Modify the pivot table to show the sum of Value instead of the mean.

Add margins to the pivot table to show the total mean for each Category and Type.

```python
df=pd.DataFrame(data={'Catégorie':['A','A','A','B','B','B'],'Type':['X','Y','X','Y','X','Y'],'Value':[1,2,3,4,5,6]})
df
```

```python
#on crée une table pivot avec comme fonction d'agregation la moyenne et value comme valeur
pivot=df.pivot_table(index='Catégorie',columns='Type',values='Value',aggfunc='mean')
pivot
```

```python
#somme a la place de la moyenne
pivot=df.pivot_table(index='Catégorie',columns='Type',values='Value',aggfunc='sum')
pivot
```

```python
# rajout des totaux avec margins=True
pivot=df.pivot_table(index='Catégorie',columns='Type',values='Value',aggfunc='sum',margins=True)
pivot
```

## Exercice 9 : Time Series Data

Create a time series DataFrame with a date range starting from '2023-01-01' for 6 periods and random values.

Set the date column as the index of the DataFrame.

Resample the data to calculate the sum for each 2-day period.

```python
#on crée un dataframe avec 365 jours à partir du 1er janvier 2023
periods=pd.date_range(start='2023-01-01',periods=265,freq='D')
periods
randomdates=np.random.choice(periods,6,replace=False)
randomdates
randomdatesdf=pd.to_datetime(randomdates)
randomdatesdf
df1=pd.DataFrame(data=randomdatesdf,columns=['Date'])
df1
```

```python
#On change l'index
df1.set_index('Date',inplace=True)
df1
```

```python
#on ressessemple pour avoir toutes les dates avec eccart
resampled_df = df1.resample('2D').sum()
resampled_df
```
